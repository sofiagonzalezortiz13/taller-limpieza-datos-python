# Punto 1: Cargar el Excel y explorar los datos

En este paso se abre el archivo Excel y se revisan sus primeras filas, tipos de datos y valores nulos.\n\n**Objetivo:** conocer la estructura de los datos antes de limpiarlos.


```python
# PUNTO 1: CARGAR EL EXCEL Y EXPLORAR LOS DATOS
import os
import unicodedata

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

ruta = r"C:\Users\User\Documents\ML\Data\datos_para_limpiar - Original.xlsx"

if not os.path.isfile(ruta):
    raise FileNotFoundError(f"No se encontró el archivo Excel: {ruta}")

hojas_disponibles = pd.ExcelFile(ruta).sheet_names
print("Hojas disponibles:", hojas_disponibles)

if "Sheet1" not in hojas_disponibles:
    raise ValueError(f"No se encontró la hoja Sheet1. Hojas disponibles: {hojas_disponibles}")

df = pd.read_excel(ruta, sheet_name="Sheet1")

columnas_requeridas = [
    "ID", "Nombre", "Edad", "Salario", "Ciudad", "Genero",
    "Nivel", "Fecha", "Ingresos", "Gastos", "Codigo", "Notas"
]
columnas_faltantes = [columna for columna in columnas_requeridas if columna not in df.columns]

if columnas_faltantes:
    raise KeyError(f"Faltan columnas requeridas en el Excel: {columnas_faltantes}")

print("Columnas encontradas:", df.columns.tolist())
print("\nPrimeras cinco filas:")
display(df.head())

print("\nInformación general y tipos de datos:")
df.info()

print("\nValores nulos por columna:")
print(df.isnull().sum())
```

# Punto 2: Eliminar las columnas ID y Notas

Aquí se quitan las columnas de identificación y notas, que no forman parte del resultado solicitado.\n\n**Objetivo:** conservar los datos útiles para el taller.


```python
# PUNTO 2: ELIMINAR LAS COLUMNAS ID Y NOTAS

df = df.drop(columns=["ID", "Notas"])
print("Columnas conservadas:", df.columns.tolist())
```

# Punto 3: Normalizar los nombres de las columnas

Se convierten los nombres de las columnas a minúsculas, se quitan tildes y se reemplazan espacios por guiones bajos.\n\n**Objetivo:** trabajar con nombres de columna uniformes y fáciles de usar en Python.


```python
# PUNTO 3: NORMALIZAR LOS NOMBRES DE LAS COLUMNAS

def quitar_tildes(texto):
    texto_normalizado = unicodedata.normalize("NFD", str(texto))
    return "".join(
        caracter
        for caracter in texto_normalizado
        if unicodedata.category(caracter) != "Mn"
    )

def normalizar_nombre_columna(nombre):
    nombre_limpio = quitar_tildes(nombre).strip().lower()
    return "_".join(nombre_limpio.split())

df.columns = [normalizar_nombre_columna(columna) for columna in df.columns]
print("Nombres de columnas normalizados:", df.columns.tolist())
```

# Punto 4: Normalizar los valores de ciudad

Se quitan espacios sobrantes y se unifican las distintas escrituras de Bogotá, Medellín y Cali.\n\n**Objetivo:** que cada ciudad tenga un solo nombre consistente.


```python
# PUNTO 4: NORMALIZAR LOS VALORES DE CIUDAD

def normalizar_ciudad(valor):
    if pd.isna(valor):
        return valor

    ciudad_limpia = " ".join(str(valor).split())
    clave = quitar_tildes(ciudad_limpia).lower()
    ciudades = {
        "bogota": "Bogotá",
        "bogot\ufffd": "Bogot\u00e1",
        "medellin": "Medellín",
        "cali": "Cali"
    }
    return ciudades.get(clave, ciudad_limpia)

df["ciudad"] = df["ciudad"].apply(normalizar_ciudad)
print(df["ciudad"].value_counts(dropna=False))
```

# Punto 5: Completar los valores nulos solicitados

Los valores faltantes de edad se completan con la media de edades válidas entre 0 y 120; salario usa la mediana, y ciudad y género usan la moda. No se imputan otras columnas.\n\n**Objetivo:** completar únicamente los nulos indicados en el taller.


```python
# PUNTO 5: COMPLETAR LOS VALORES NULOS SEGÚN LAS REGLAS

edades_numericas = pd.to_numeric(df["edad"], errors="coerce")
edades_validas = edades_numericas.between(0, 120)

if edades_numericas.isna().any():
    if not edades_validas.any():
        raise ValueError("No hay edades válidas entre 0 y 120 para calcular la media.")
    media_edad_valida = edades_numericas.loc[edades_validas].mean()
    df["edad"] = edades_numericas.fillna(media_edad_valida)
else:
    df["edad"] = edades_numericas

# La mediana se calcula con los salarios numéricos disponibles.
df["salario"] = pd.to_numeric(df["salario"], errors="raise")
if df["salario"].isna().any():
    mediana_salario = df["salario"].median()
    if pd.isna(mediana_salario):
        raise ValueError("No hay salarios válidos para calcular la mediana.")
    df["salario"] = df["salario"].fillna(mediana_salario)

# La ciudad y el género se completan con su valor más frecuente.
for columna in ["ciudad", "genero"]:
    if df[columna].isna().any():
        modas = df[columna].mode(dropna=True)
        if modas.empty:
            raise ValueError(f"No hay valores válidos para calcular la moda de {columna}.")
        df[columna] = df[columna].fillna(modas.iloc[0])

print("Valores nulos después del tratamiento:")
print(df[["edad", "salario", "ciudad", "genero"]].isna().sum())
```

# Punto 6: Convertir edad y fecha a sus tipos correctos

Edad se convierte a valor numérico y fecha a tipo datetime.\n\n**Objetivo:** dejar esas columnas preparadas para cálculos y validaciones.


```python
# PUNTO 6: CONVERTIR EDAD A NÚMERO Y FECHA A DATETIME

df["edad"] = pd.to_numeric(df["edad"], errors="raise")
fechas_convertidas = pd.to_datetime(df["fecha"], errors="coerce")
fechas_invalidas = df["fecha"].notna() & fechas_convertidas.isna()

if fechas_invalidas.any():
    ejemplos = df.loc[fechas_invalidas, "fecha"].head().tolist()
    raise ValueError(f"Hay fechas que no se pudieron convertir: {ejemplos}")

df["fecha"] = fechas_convertidas
print(df[["edad", "fecha"]].dtypes)
```

# Punto 7: Eliminar edades y salarios fuera de los límites

Antes de escalar, se eliminan filas con edad menor que 0, mayor que 120 o salario superior a 1.000.000.000.\n\n**Objetivo:** retirar registros fuera de los límites definidos usando los valores originales.


```python
# PUNTO 7: ELIMINAR EDADES Y SALARIOS FUERA DE LOS LÍMITES

# Este filtro se aplica antes del escalado, con las edades originales.
edades_fuera_rango = ~df["edad"].between(0, 120)
salarios_superiores_limite = df["salario"] > 1_000_000_000
filas_para_eliminar = edades_fuera_rango | salarios_superiores_limite

print("Filas con edad menor que 0 o mayor que 120:", int(edades_fuera_rango.sum()))
print("Filas con salario superior a 1.000.000.000:", int(salarios_superiores_limite.sum()))
print("Filas eliminadas por al menos una regla:", int(filas_para_eliminar.sum()))

df = df.loc[~filas_para_eliminar].copy()

if not df["edad"].between(0, 120).all():
    raise AssertionError("Quedaron edades fuera del rango 0–120.")
if not df["salario"].le(1_000_000_000).all():
    raise AssertionError("Quedaron salarios superiores a 1.000.000.000.")

print("Registros válidos antes del escalado:", len(df))
```

# Punto 8: Codificar el nivel de forma ordinal

Los niveles bajo, medio y alto se convierten en 1, 2 y 3. Los valores vacíos y n/a permanecen como nulos.\n\n**Objetivo:** representar el nivel como una escala ordenada.


```python
# PUNTO 8: CODIFICAR NIVEL Y CONSERVAR VACÍOS Y N/A COMO NULOS

nivel_limpio = df["nivel"].astype("string").str.strip().str.lower()
nivel_limpio = nivel_limpio.mask(nivel_limpio.isin(["", "n/a"]))
mapa_nivel = {"bajo": 1, "medio": 2, "alto": 3}
valores_no_reconocidos = nivel_limpio.notna() & ~nivel_limpio.isin(mapa_nivel)

if valores_no_reconocidos.any():
    valores = nivel_limpio.loc[valores_no_reconocidos].unique().tolist()
    raise ValueError(f"Hay valores de nivel no contemplados: {valores}")

df["nivel"] = nivel_limpio.map(mapa_nivel).astype("Int64")
print(df["nivel"].value_counts(dropna=False))
```

# Punto 9: Codificar género y ciudad con One-Hot Encoding

Genero y ciudad se convierten en columnas indicadoras con `pd.get_dummies` y `drop_first=True`.\n\n**Objetivo:** preparar esas categorías para análisis numérico.


```python
# PUNTO 9: APLICAR ONE-HOT ENCODING A GÉNERO Y CIUDAD

# Unificamos las variantes observadas antes de crear las columnas indicadoras.
generos_canonicos = {
    "femenino": "Femenino",
    "masculino": "Masculino",
    "otro": "Otro"
}
df["genero"] = df["genero"].apply(
    lambda valor: generos_canonicos.get(
        str(valor).strip().casefold(), str(valor).strip()
    ) if pd.notna(valor) else valor
)
df = pd.get_dummies(
    df,
    columns=["genero", "ciudad"],
    drop_first=True,
    dtype=int
)
df.columns = [
    "_".join(quitar_tildes(columna).strip().lower().split())
    for columna in df.columns
]
print("Columnas después de la codificación:", df.columns.tolist())
```

# Punto 10: Escalar edad e ingresos

MinMaxScaler transforma edad e ingresos a un rango entre 0 y 1 y conserva sus valores nulos. También se guardan copias de ingresos y gastos originales para calcular la utilidad correctamente.\n\n**Objetivo:** escalar las variables sin perder los montos necesarios para la operación monetaria.


```python
# PUNTO 10: ESCALAR EDAD E INGRESOS ENTRE 0 Y 1

# Guardamos los montos originales para calcular utilidad después.
df["ingresos"] = pd.to_numeric(df["ingresos"], errors="raise")
df["gastos"] = pd.to_numeric(df["gastos"], errors="raise")
ingresos_originales = df["ingresos"].copy()
gastos_originales = df["gastos"].copy()

for columna in ["edad", "ingresos"]:
    valores_presentes = df[columna].notna()
    if valores_presentes.any():
        escalador = MinMaxScaler()
        df.loc[valores_presentes, columna] = escalador.fit_transform(
            df.loc[valores_presentes, [columna]]
        ).ravel()

print(df[["edad", "ingresos"]].head())
```

# Punto 11: Eliminar filas duplicadas

Se eliminan filas que estén duplicadas en el DataFrame.\n\n**Objetivo:** evitar contar más de una vez registros repetidos.


```python
# PUNTO 11: ELIMINAR LAS FILAS DUPLICADAS

cantidad_duplicados = df.duplicated().sum()
df = df.drop_duplicates().copy()
print("Filas duplicadas eliminadas:", cantidad_duplicados)
print("Registros restantes:", len(df))
```

# Punto 12: Calcular utilidad con valores originales

La utilidad se calcula restando los gastos originales a los ingresos originales, no los ingresos escalados.\n\n**Objetivo:** conservar el sentido monetario de la resta.


```python
# PUNTO 12: CREAR UTILIDAD CON INGRESOS Y GASTOS ORIGINALES

# Se usan las copias guardadas antes de escalar ingresos.
df["utilidad"] = (
    ingresos_originales.reindex(df.index)
    - gastos_originales.reindex(df.index)
)
print(df[["ingresos", "gastos", "utilidad"]].head())
```

# Punto 13: Exportar y verificar el archivo CSV

El DataFrame final se guarda como `datos_limpios.csv` sin índice y luego se comprueban el archivo, las columnas, los duplicados y la cantidad de filas.\n\n**Objetivo:** entregar un CSV verificable con el resultado de la limpieza.


```python
# PUNTO 13: EXPORTAR Y VERIFICAR EL ARCHIVO FINAL

# Verificamos los datos transformados antes de guardar el archivo.
if not pd.api.types.is_datetime64_any_dtype(df["fecha"]):
    raise AssertionError("La columna fecha no tiene un tipo datetime.")
if not df["edad"].dropna().between(0, 1).all():
    raise AssertionError("La edad escalada no está entre 0 y 1.")
if not df["ingresos"].dropna().between(0, 1).all():
    raise AssertionError("Los ingresos escalados no están entre 0 y 1.")
if not df["utilidad"].equals(
    ingresos_originales.reindex(df.index) - gastos_originales.reindex(df.index)
):
    raise AssertionError("La utilidad no corresponde a los valores monetarios originales.")

carpeta_proyecto = os.path.dirname(os.path.dirname(ruta))
ruta_salida = os.path.join(carpeta_proyecto, "datos_limpios.csv")
df.to_csv(ruta_salida, index=False, encoding="utf-8")

if not os.path.isfile(ruta_salida):
    raise FileNotFoundError(f"No se creó el archivo esperado: {ruta_salida}")

datos_exportados = pd.read_csv(ruta_salida)

if "id" in datos_exportados.columns or "notas" in datos_exportados.columns:
    raise AssertionError("El CSV todavía contiene ID o Notas.")
if "nombre" not in datos_exportados.columns or "codigo" not in datos_exportados.columns:
    raise AssertionError("El CSV debe conservar Nombre y Codigo.")
if datos_exportados.duplicated().any():
    raise AssertionError("El CSV contiene filas duplicadas.")
if "Unnamed: 0" in datos_exportados.columns:
    raise AssertionError("El CSV contiene una columna de índice no solicitada.")
if len(datos_exportados) != len(df):
    raise AssertionError("El número de filas exportadas no coincide con el DataFrame.")

print("Archivo creado y verificado:", ruta_salida)
print("Primeras filas del resultado:")
display(datos_exportados.head())
print("\nTipo de fecha en el DataFrame:", df["fecha"].dtype)
print("\nTipos de datos del resultado leído:")
print(datos_exportados.dtypes)
print("\nValores nulos restantes por columna:")
print(datos_exportados.isna().sum())
print("\nCantidad final de registros:", len(datos_exportados))
```
